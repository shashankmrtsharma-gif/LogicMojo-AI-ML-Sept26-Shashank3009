**Loop Control Statment- Break, Pass and Continue**

```python
for i in range(1,6):
   if i == 3:
       print(i)
       break # print i == 3 this is the break point
```

```python
for i in range(1,6):
    if i == 3:
        break
    print(i) # break and print range till the break point
```

```python
for i in range(1,6):
    if i == 3:
        continue
    print(i) # except condition and excluded range 
```

```python
for i in range(1,6):
    if i == 3:
        pass
    print(i) # except exclude 
```

##**Scenario**: You are building a simple student result management system. Write a Python program that performs the following tasks:
1. Ask the user to enter marks for 5 students using a while loop.
2. If the entered marks are outside the range of 0–100, display an invalid marks message and skip that entry using continue. The user must still enter marks until 5 valid entries have been collected.
3. Use a for loop to process the collected marks.
4. If marks are 90 or above, print Grade A.
5. If marks are between 75 and 89, print Grade B.
6. If marks are between 60 and 74, print Grade C.
7. If marks are between 40 and 59, print Pass.
8. Otherwise, print Fail.
9. If marks are exactly 0, use break to stop processing further students.
10. Include a pass statement as a placeholder for a future feature, such as saving student results to a file.
11. Finally, print how many valid marks were collected

```python
print ("student result management system")
print ("-"*50)
i = 0
while i  < 5 :
    student_marks=input(f"Enter the student name {i + 1}:")
    marks=float(student_marks)
    if marks < 0 or marks > 100:
        print("invalid marks message")
        continue
    if marks >= 90:
        print("Garde A")
    elif marks >= 75:
        print("Garde B")
    elif marks >= 60:
        print("Garde C")
    elif marks >= 40:
        print("Pass")
    else:
        print("Fail")
    if marks == 0:
        break             ##Need to check why 4 count "?
    pass # "placeholder for future feature"
    
    i= i + 1

print(f"total valid marks collected {i}")
```
