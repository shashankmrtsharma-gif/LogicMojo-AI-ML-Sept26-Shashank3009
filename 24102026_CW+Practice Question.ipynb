```python
age_input = input("Enter an age: ")
name = input("Enter a name: ")

collected_errors = []

try:
    if age_input == "":
        collected_errors.append(ValueError("age cannot be empty"))
    else:
        try:
            age = int(age_input)
            if age < 0:
                collected_errors.append(ValueError("age cannot be negative"))
        except ValueError:
            collected_errors.append(ValueError("age must be a valid number"))

    # Validate Name
    if name == "":
        collected_errors.append(NameError("Username cannot be empty."))

    # 2. If errors were found, raise them ALL simultaneously
    if collected_errors:
        raise ExceptionGroup("Validation Failed", collected_errors)

# 3. Use except* to handle multiple errors at once
except* ValueError as s:
    for error in s.exceptions:
        print(f"Caught ValueError: {error}")

except* NameError as s:
    for error in s.exceptions:
        print(f"Caught NameError: {error}")
```

```python
#Q2. Positive Number Checker
# Ask the user to enter a number. 
# Handle invalid numeric input using try-except. 
#If the number is zero or negative, raise a ValueError. 
#If it is positive, check whether it is even or odd and display the result.

try:
    num = int(input("Enter a number: "))
    
    if num <= 0:
        raise ValueError("pls enter a positive number")

    if num % 2 == 0:
        print("No. is even")
    else: 
        print("No. is odd")

except ValueError as s:
    print(s)
```

```python
#Q4. Division Calculator
#Take two numbers as input. Handle invalid input using try-except. 
#If the second number is zero, raise a ZeroDivisionError. Otherwise, divide the first number by the second and print the result. 
#Use else to display the result only when no exception occurs.
try:
    a = int(input("Enter the first number"))
    b = int(input("Enter the second number"))

    if b == 0:
        raise ZeroDivisionError("Number can't be zero")

except ValueError:
    print("Invalid input! Please enter valid integers.")

except ZeroDivisionError as s:
    print(s)

else:
    result = a / b
    print(result)
```

```python
# Q5. Student Marks Validation
#Take marks as input. If the input is not a valid integer, handle the exception. 
#If marks are less than 0 or greater than 100, raise a ValueError. 
#Otherwise, use if-else to print "Pass" if marks are at least 40 and "Fail" otherwise.

try:
    marks = int(input("enter the marks"))
    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100.")
        
except ValueError as s:
    if "between 0 and 100" in str(s):
        print(s)
    else:
        print("Invalid input! Please enter a valid integer.")

else:
    if marks >= 40:
        print("Pass")
    else:
        print("Fail")


```

```python
#Q6. Create a Student File
#Use "x" mode to create a file named student.txt. 
#Take the student's name, age, and course as input and write these details into the file. 
#Handle FileExistsError if the file already exists.


try:
    file = open("student.txt", "x")

    name = input("Enter student name: ")
    age = input("Enter student age: ")
    course = input("Enter student course: ")

    file.write("Student Name: " + name + "\n")
    file.write("Age: " + age + "\n")
    file.write("Course: " + course + "\n")

    file.close()

    print("Student details saved successfully.")

except FileExistsError:
    print("Error: student.txt already exists.")
```

```python
#Q7. Write Student Records
#Use "w" mode to create or overwrite students.txt. 
#Take the details of three students and write each student's name and marks on a separate line. 
#Handle any appropriate input errors using try-except.

try:
    file = open("students.txt", "w")

    for i in range(1, 4):
        name = input(f"Enter student name{i}:")
        marks = float(input(f"Enter student marks{i}: "))

        file.write(name + " - " + str(marks) + "\n")

    file.close()

    print("All student records saved successfully.")

except ValueError:
    print("Error: Please enter valid numeric marks.")
                     
```

```python
#Q8. Append Student Records
#Use "a" mode to add a new student's name and marks to the existing students.txt file. 
#If the marks are below 0 or above 100, raise a ValueError. Handle the exception without adding invalid records to the file.

try:
    name = input("Enter student name: ")
    marks = float(input("Enter student marks: "))

    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100.")

    with open("students.txt", "a") as file:
        file.write(name + " - " + str(marks) + "\n")

    print("Student record added successfully.")

except ValueError as s:
    print(s)
file.close()
```

```python
#Q9. Read a File Safely
#Ask the user to enter a filename. Open the file using "r" mode and display its contents. 
#Handle FileNotFoundError if the file does not exist. If the file is empty, print "File is empty".

try:
    file = open(input("Enter the file name"),"r")
    content = file.read()
    if content == "":
        print("file is empty")
    else:
        print(content)
    
    file.close()
except FileNotFoundError:
    print("file does not exist")
```

```python
#Q10. Number Storage in a File
#Take five numbers from the user and write them to numbers.txt using "w" mode. If a user enters an invalid number, handle the exception. If a number is negative, raise a ValueError. Store only valid, non-negative numbers in the file.
file = open("numbers.txt", "w")

for i in range(1, 6):
    try:
        numbers = int(input(f"Enter the num {i}: "))

        if numbers < 0:
            raise ValueError("Invalid Number")

        file.write(str(numbers) + "\n")

    except ValueError as s:
        print(s)

file.close()
```

```python
#Q11. Search for a Student
#Read students.txt using "r" mode. Ask the user to enter a student name. 
#If the name exists in the file, print "Student found"; otherwise, print "Student not found". Handle FileNotFoundError if the file does not exist.
try: 
    file = open("students.txt","r")
    content=file.read()
    name = input("enter the student name")
    if name in content:
        print("student found")
    else:
        print("student not found")

except FileNotFoundError:
    print("FileNotFoundError")
    
```

```python
#Q12. File Creation with Validation
#Ask the user to enter a filename. If the filename is empty, 
#raise a ValueError. Otherwise, use "x" mode to create the file. Handle both ValueError and FileExistsError appropriately.

try:
    filename = input("Enter filename: ")

    if filename == "":
        raise ValueError("Filename cannot be empty")

    file = open(filename, "x")
    file.close()

    print("File created successfully")

except ValueError as s:
    print(s)

except FileExistsError:
    print("File already exists")
```

```python
#Q13. Append Only Passing Students
#Take a student's name and marks as input. If marks are outside the range 0-100, raise a ValueError. 
#If marks are 40 or above, append the student's details to passed_students.txt using "a" mode. Otherwise, print "Student failed; record not added".

try :
    name = input("enter the student name")
    marks = int(input("enter the marks"))

    if marks < 0 and marks > 100:
        raise ValueError("Marks are outside the range 0-100")
    if marks >= 40:
        with open("passed_students.txt","a") as file:
            file.write(name + " - " + str(marks) + "\n")
    else:
        print("Student failed")
        
except ValueError as s:
    print(s)
```

```python
#Q14. Count Lines in a File
# Ask the user for a filename and read it using "r" mode. 
#Count the number of non-empty lines in the file. Handle FileNotFoundError. If the file contains no non-empty lines, print "No records found".

try :
    file = open(input("Enter the filename: "), "r")
    count = 0
    
    for line in file:
            if line.strip() != "":
                count += 1
                
    if count == 0:
        print("No records found")
    else:
        print("Number of non-empty lines:", count)

except FileNotFoundError:
        print("File does not exist")
  
```
