```python
# Q39. A dataset contains the following actual customer
# categories:
#
# actual_categories = ["A", "B", "A", "C", "B", "A", "C", "B"]
#
# Create a dictionary that counts the frequency of each
# category.
#
# Then calculate the percentage of records belonging to
# each category.
#
# Example:
# Category A: 37.5%
#
# Print both the counts and percentages.


actual_categories = ["A", "B", "A", "C", "B", "A", "C", "B"]

count={}
for categories in actual_categories:
    if categories in count:
        count[categories] += 1
    else:
        count[categories] = 1
print(count)

for category, category_count in count.items():
    percentage = (category_count / len(actual_categories)) * 100  # Added fixed line
    print(f"Category {category}: {percentage:.1f}%")
```

```python
# Q40. A data science team maintains two dictionaries
# containing model performance results:
#
# experiment_1 = {
#     "Linear Regression": 0.82,
#     "Decision Tree": 0.88,
#     "Random Forest": 0.94
# }
#
# experiment_2 = {
#     "Decision Tree": 0.90,
#     "Random Forest": 0.96,
#     "KNN": 0.85
# }
#
# Perform the following tasks:
#
# 1. Merge both dictionaries into a new dictionary.
# 2. If a model appears in both dictionaries, retain the
#    higher accuracy score.
# 3. Print the final dictionary.
# 4. Identify the model with the highest accuracy.
# 5. Print all models whose final accuracy is greater than 0.90.


experiment_1 = {"Linear Regression": 0.82,"Decision Tree": 0.88,"Random Forest": 0.94}
experiment_2 = {"Decision Tree": 0.90,"Random Forest": 0.96,"KNN": 0.85}

new_experiment= experiment_1 | experiment_2
print(f"merge_dict : {new_experiment}")

highest_score =0
for model, score in new_experiment.items():
    if score >= highest_score:
        highest_score = score  
        best_model = model 
print(best_model,highest_score)

for model, score in new_experiment.items():
    if score > 0.90:
        print(f"{model}: {score:.2f}")
```
