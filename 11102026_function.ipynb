```python
def multiply(a, b):
    return a * b

def summ(a, b, operator):
    if operator == "*":
        return multiply(a, b)  # Added 'return' here
    else:
        return a + b

# Testing the code
print(summ(1, 2, "*"))  # Output: 2
print(summ(1, 2, "+"))  # Output: 3
```

```python
import math

def multiply(*args):
    # args becomes a tuple of all numbers passed in
    return math.prod(args)

def abc(*num):
    # Unpack the tuple and pass it directly to multiply
    return multiply(*num)

# Now it works with ANY number of inputs dynamically!
print(abc(5, 10))        # Output: 50  (5 * 10)
print(abc(2, 3, 4))      # Output: 24  (2 * 3 * 4)
print(abc(1, 2, 3, 4, 5))# Output: 120 (1 * 2 * 3 * 4 * 5)
```

```python
def add(a,b):
    return a+b
def multiply(a,b):
    return a*b
def subtract(a,b):
    return a-b
def divide(a,b):
    return a/b

operation_map = {
    "+" : add,
    "-" : subtract,
    "*" : multiply,
    "/" : divide}

def calculate (a,b,operator):
    if operator in operation_map:
        selected_function = operation_map[operator]
        return selected_function(a,b)
    else:
        return "unknow"
calculate(4,2,"/")
```

```python
operation_map = {
    "+" : lambda a,b: a+b,
    "-" : lambda a,b: a-b,
    "*" : lambda a,b: a*b,
    "/" : lambda a,b: a/b}

def calculate (a,b,operator):
    if operator in operation_map:
        selected_function = operation_map[operator]
        return selected_function(a,b)
    else:
        return "unknow"
calculate(4,2,"/")
```
