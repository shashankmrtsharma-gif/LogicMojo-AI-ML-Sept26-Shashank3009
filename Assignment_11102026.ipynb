```python
# 60. Build a mini training simulation using functions, a decorator, a loop, conditions, break, and a lambda function. 
#Print the loss for each epoch, stop when loss reaches 0.10 or lower, and return the final loss. Use only the Python concepts covered in class.



reduce_loss = lambda loss: round(loss - 0.10, 2)

def training_decorator(func):
    def wrapper():
        print("Training Started")
        result = func()
        print("Training Finished")
        return result
    return wrapper

# Function to simulate training
@training_decorator
def train_model():
    loss = 0.8

    for epoch in range(1, 11):
        print(f"Epoch {epoch}, Loss: {loss:.2f}")

        if loss <= 0.10:
            break

        loss = reduce_loss(loss)

    return loss

final_loss = train_model()
print("Final Loss:", final_loss)
```
