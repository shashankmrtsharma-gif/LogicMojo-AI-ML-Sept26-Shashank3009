```python

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Student name:", self.name)
        print("Marks:", self.marks)

    def check_result(self):
        if self.marks >= 40:
            print("Pass")
        else:
            print("FAIL")

    def show_details(self):
        self.display()
        self.check_result()
        print("----------------")

student1 = Student("Shafi", 95)
student2 = Student("Surja", 60)

student1.show_details()
student2.show_details()

```

```python

class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show_details(self):
        print("Employee Name:", self.name)
        print("Salary:", self.salary)


class DataScientist(Employee):

    def work(self):
        print("Working on Machine Learning models")


# Create object of child class
emp1 = DataScientist("Rahul", 80000)

# Call parent class method
emp1.show_details()

# Call child class method
emp1.work()
```

```python

class Python:
    def python_skill(self):
        print("Knows Python")


class SQL:
    def sql_skill(self):
        print("Knows SQL")


# Multiple Inheritance
class DataScientist(Python, SQL):
    pass


# Create object
person = DataScientist()

# Call methods from both parent classes
person.python_skill()
person.sql_skill()
```

```python
class Dog:
    def sound(self):
        print("Barks")

class cat:
    def sound(self):
        print("Meow")

class cow:
    def sound(self):
        print("moos")

animals = [Dog(), cat(), cow()]

for animal in animals:
    animal.sound()
```

```python

# Parent Class
class Employee:
    def employee_details(self):
        print("Employee belongs to the company")


# Hierarchical Inheritance
class Developer(Employee):
    def write_code(self):
        print("Developer writes Python code")


class Manager(Employee):
    def manage_team(self):
        print("Manager manages the team")


# Multilevel Inheritance
class SeniorDeveloper(Developer):
    def review_code(self):
        print("Senior Developer reviews code")


# Hybrid Inheritance (Multiple Inheritance)
class TechnicalLead(SeniorDeveloper, Manager):
    def lead_project(self):
        print("Technical Lead manages the project")


# Create object
person = TechnicalLead()

# Call methods from all inherited classes
person.employee_details()
person.write_code()
person.review_code()
person.manage_team()
person.lead_project()
```
