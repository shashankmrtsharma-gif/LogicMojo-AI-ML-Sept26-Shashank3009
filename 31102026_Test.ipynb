```python
# Write a Python program to count the number of vowels in a given string without using external libraries.
def count_vowel(input_string):
    vowels = {"a", "e", "i", "o", "u", "A", "E", "I", "O", "U"}
    count = 0

    for char in input_string:
        if char in vowels:
            count += 1
    return count
input_string = input("enter the string")
print(f"Number of vowels: {count_vowel(input_string)}")
```

```python
# Create a function named calculate_discount() 
# that accepts the original price and discount percentage and returns the final price after applying the discount.

def calculate_discount(original_price, discount_percentage):
    discount_amount = original_price * (discount_percentage / 100)
    final_price = original_price - discount_amount
    return final_price

price = float(input("Enter the original price: "))
discount = float(input("Enter the discount percentage: "))

saved_price = calculate_discount(price, discount)
print(f"final price after a {discount}% discount is: {saved_price:.2f}")
```

```python
# Write a Python program to check whether a given string is a palindrome without using the built-in reverse function.Example:Input: madam Output: Palindrome

def is_palindrome(text):
    text = text.lower()
    n = len(text)

    for i in range(n // 2):
        if text[i] != text[n - 1 - i]:
            return "Not a Palindrome"
    return "Palindrome"

user_input = input("Enter a string: ")
print(is_palindrome(user_input))
```

```python
# Write a program to find the second-largest unique number in a list without using the sort() or sorted() functions.Input: [10, 20, 40, 40, 30, 50] Output: 40

def find_second_largest(numbers):
    largest = float()
    second_largest = float()
    
    for num in numbers:
        if num > largest:
            second_largest = largest
            largest = num
        elif num > second_largest and num != largest:
            second_largest = num
    return second_largest if second_largest != float() else None

input_list = [10, 20, 40, 40, 30, 50]
output = find_second_largest(input_list)
print(f"Input: {input_list}")
print(f"Output: {output}")
```

```python
# Q19. Write a Python function that accepts a list of numbers and returns a dictionary containing the count of even numbers, odd numbers, and zeros.Example:Input: [0, 1, 2, 3, 4, 5, 0]#Output:
#{
    #'even': 2,
    #'odd': 3,
    #'zero': 2
#}

def count_numbers(numbers):
    counts = {
        'even': 0,
        'odd': 0,
        'zero': 0
    }
    
    for num in numbers:
        if num == 0:
            counts['zero'] += 1
        elif num % 2 == 0:
            counts['even'] += 1
        else:
            counts['odd'] += 1
            
    return counts

input_list = [0, 1, 2, 3, 4, 5, 0]
output = count_numbers(input_list)
print(output)
```

```python
# You are developing a simple banking application.
#Write a function withdraw() that:
#•	Accepts account balance and withdrawal amount.
#•	Raises an error if the withdrawal amount is negative.
#•	Raises an error if the withdrawal amount exceeds the available balance.
#•	Returns the updated balance if the transaction is successful.
#Use appropriate exception handling.


def withdraw(balance, amount):
    if amount < 0:
        raise ValueError("Withdrawal amount cannot be negative.")
    if amount > balance:
        raise ValueError("Insufficient funds.")
    
    return balance - amount
try:
    updated_balance = withdraw(balance=5000, amount=2000)
    print(f"Success! New Balance: {updated_balance}")
except ValueError as e:
    print(e)
```

```python
#Write a Python program to read a CSV file named employees.csv containing the following columns:
#Employee_ID,Name,Department,Salary
#Calculate the average salary department-wise and display the department with the highest average salary.

import csv

file = open("employee.csv", "r")
data = list(csv.DictReader(file))
file.close()

department_salary = {}
department_count = {}

for employee in data:
    department = employee["Department"]
    salary = float(employee["Salary"])

    if department not in department_salary:
        department_salary[department] = 0
        department_count[department] = 0

    department_salary[department] += salary
    department_count[department] += 1

average_salary = {}

for department in department_salary:
    average_salary[department] = department_salary[department] / department_count[department]
    print(department, average_salary[department])

highest_department = max(average_salary, key=average_salary.get)

print("Department with highest average salary:", highest_department)
print("Highest average salary:", average_salary[highest_department])
```

```python
#Write a Python decorator named execution_time that measures how long a function takes to execute.
#Apply this decorator to a function that calculates the sum of numbers from 1 to 1,000,000.

import time

def execution_time(func):
    def wrapper():
        start_time = time.time()

        result = func()

        end_time = time.time()

        print("Execution time:", end_time - start_time, "seconds")

        return result

    return wrapper


@execution_time
def calculate_sum():
    total = 0

    for i in range(1, 1000001):
        total += i

    print("Sum:", total)


calculate_sum()
```

```python
#Create a function named login() that accepts any number of keyword arguments using **kwargs.
#The function should:
#•	Accept username and password.
#•	Validate whether both are provided.
#•	Display the remaining optional parameters.
#•	Return a success or failure message.

def login(**kwargs):
    if "username" not in kwargs or "password" not in kwargs:
        return "Failure: Both username and password must required."
    
    print("Optional parameters are:")
    for key, value in kwargs.items():
        if key != "username" and key != "password":
            print(f"- {key}: {value}")
            
    return "Success: Logged in successfully!"


status = login(username="sam", password="123", role="readonly", age=18)
print("-" * 30)
print(login(username="shashank", age=18))

print("-" * 30)

status = login(username="shashank", password="123", role="admin", age=18)
print(status)
```

```python
#Given the following list of dictionaries:#Write a program to group employees department-wise and calculate the total salary for each department.
employees = [
    {"name": "Amit", "salary": 50000, "department": "IT"},
    {"name": "Neha", "salary": 70000, "department": "HR"},
    {"name": "Raj", "salary": 60000, "department": "IT"},
    {"name": "Priya", "salary": 80000, "department": "HR"}
]


department_employees = {}
department_salaries = {}

for emp in employees:
    dept = emp["department"]
    name = emp["name"]
    salary = emp["salary"]

    if dept not in department_employees:
        department_employees[dept] = []
        department_salaries[dept] = 0
    
    department_employees[dept].append(name)
    department_salaries[dept] += salary

print("--- Department-wise Groups ---")
for dept, names in department_employees.items():
    print(f"{dept} Department: {', '.join(names)}")

print("\n--- Total Salaries ---")
for dept, total_salary in department_salaries.items():
    print(f"{dept} Department Total Salary: Rs {total_salary}")
```

```python
#You are developing a simple caching system.
#Implement a function that stores previously calculated results in a dictionary. If the same input is received again, return the cached result instead of calculating it again.
#Demonstrate your implementation using a function that calculates the square of a number.

cache_memory = {}
def square (number):
    if number in cache_memory:
        print(f"{number} in memory")
        return cache_memory[number]

    print(f"calculating square of {number}")
    result = number * number 
        
    cache_memory[number] = result
    return result

print(f"result: {square(5)}")
print(f"result: {square(10)}")
print(f"current cache content: {cache_memory}")
```

```python
# Write a Python program that accepts a list of transactions:
transactions = [
    {"id": 1, "amount": 500, "status": "success"},
    {"id": 2, "amount": 300, "status": "failed"},
    {"id": 3, "amount": 700, "status": "success"},
    {"id": 4, "amount": 200, "status": "success"}
]
#Calculate:
#•	Total successful transaction amount.
#•	Number of failed transactions.
#•	Average amount of successful transactions.

total_success_amount = 0
success_count = 0
failed_count = 0

for tx in transactions:
    if tx["status"] == "success":
        total_success_amount += tx["amount"]
    elif tx["status"] == "failed":
        failed_count += 1
if success_count > 0:
    avg_success_amount = total_success_amount / success_count
else:
    avg_success_amount = 0


print(f"Total successful transaction amount: Rs {total_success_amount}")
print(f"Number of failed transactions: {failed_count}")
print(f"Average amount of successful transactions: Rs {avg_success_amount}")
```

```python
#You are building a file-processing application.
#Write a Python program that:
#•	Opens a text file.
#•	Counts the total number of lines, words, and characters.
#•	Handles the situation where the file does not exist.
#•	Ensures the file is properly closed even if an exception occurs.

try :
    file=open("students.txt","r")
    count_lines = 0
    count_words = 0
    count_char = 0
    for line in file:
        count_lines += 1
        count_char += len(line)
        words = line.split()
        count_words += len(words)
    if count_lines == 0:
            print("No records found")
    else:
        print(f"Number of lines: {count_lines}")
        print(f"Number of words: {count_words}")
        print(f"Number of characters: {count_char}")
        
except FileNotFoundError:
    print("file does not exist")
except Exception as e:
    print(e)
if 'file' in locals():
    file.close()
    print("File closed safely.")
```
